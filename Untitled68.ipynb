```python

# Install dependencies (Chỉ cần chạy 1 lần ở phiên làm việc)
!pip -q install pandas numpy pytz requests lxml openpyxl playwright yfinance
!playwright install chromium
!apt-get install -y libatk-adaptor libxcomposite1 --no-install-recommends

import re
import io
import numpy as np
import pandas as pd
import requests
import yfinance as yf
from datetime import datetime, timezone, timedelta
import pytz
from lxml import html
import asyncio
import nest_asyncio
from playwright.async_api import async_playwright

nest_asyncio.apply()

# Constants
TROY_OZ_PER_TONNE = 32150.7466
VN_TZ = pytz.timezone("Asia/Ho_Chi_Minh")
NY_TZ = pytz.timezone("America/New_York")
UA = "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122 Safari/537.36"
ARCHIVE_XLSX_URL = "https://api.spdrgoldshares.com/api/v1/historical-archive?exchange=NYSE&lang=en&product=gld"
SPDR_GLD_PAGE = "https://www.spdrgoldshares.com/usa/gld/"

GLD_ANNUAL_FEE = 0.0040  # 0.40% management fee
NYSE_MARKET_CLOSE_HOUR = 16  # 4:00 PM ET

# =======================
# TIMEZONE & TIME FUNCTIONS
# =======================
def convert_ny_date_to_vn_str(ny_date_str, market_close_hour=NYSE_MARKET_CLOSE_HOUR):
    ny_date = pd.to_datetime(ny_date_str).date()
    ny_datetime = NY_TZ.localize(datetime.combine(ny_date, datetime.min.time().replace(hour=market_close_hour)))
    return ny_datetime.astimezone(VN_TZ).strftime("%d/%m/%Y %H:%M")

def get_official_effective_utc(date_str, market_close_hour=NYSE_MARKET_CLOSE_HOUR):
    ny_date = pd.to_datetime(date_str).date()
    ny_datetime = NY_TZ.localize(datetime.combine(ny_date, datetime.min.time().replace(hour=market_close_hour)))
    return ny_datetime.astimezone(timezone.utc)

# =======================
# CALCULATION FUNCTIONS
# =======================
def tonnes_to_usd(tonnes, price_per_oz): return tonnes * TROY_OZ_PER_TONNE * price_per_oz

def calculate_total_fund_value(official_tonnes, gold_price):
    total_ounces = official_tonnes * TROY_OZ_PER_TONNE
    total_value_usd = total_ounces * gold_price
    return {"total_tonnes": official_tonnes, "total_ounces": total_ounces, "total_value_usd": total_value_usd, "gold_price_used": gold_price}

def calculate_dynamic_threshold(official_tonnes, gold_price):
    fund_value = calculate_total_fund_value(official_tonnes, gold_price)
    total_value_usd = fund_value["total_value_usd"]
    thresholds = {
        "noise_filter": {"percent": 0.05, "tonnes": official_tonnes * 0.0005, "usd": total_value_usd * 0.0005},
        "weak_signal": {"percent": 0.15, "tonnes": official_tonnes * 0.0015, "usd": total_value_usd * 0.0015},
        "moderate_signal": {"percent": 0.4, "tonnes": official_tonnes * 0.004, "usd": total_value_usd * 0.004}
    }
    print(f"💰 TỔNG TÀI SẢN QUỸ GLD: {fund_value['total_value_usd']/1e9:.1f} Tỷ USD ({fund_value['total_tonnes']:,.1f} tonnes)")
    print(f"📊 THRESHOLD ĐỘNG: Noise={thresholds['noise_filter']['usd']/1e6:.0f}M | Weak={thresholds['weak_signal']['usd']/1e6:.0f}M | Strong={thresholds['moderate_signal']['usd']/1e6:.0f}M")
    return thresholds, fund_value

def classify_flow_advanced(flow_tonnes, thresholds):
    if not np.isfinite(flow_tonnes): return "❓ KHÔNG ĐỦ DỮ LIỆU"
    abs_flow = abs(flow_tonnes)
    direction = "BUY 📈" if flow_tonnes > 0 else "SELL 📉"
    if abs_flow < thresholds["noise_filter"]["tonnes"]: return f"⚪ NOISE/ARBITRAGE"
    elif abs_flow < thresholds["weak_signal"]["tonnes"]: return f"🟡 WEAK {direction}"
    elif abs_flow < thresholds["moderate_signal"]["tonnes"]: return f"🟠 MODERATE {direction}"
    else: return f"🔥 STRONG {direction}"

def calculate_official_flow_with_correct_fee(official):
    if not np.isfinite(official.get("delta_official_tonnes_vs_prev", np.nan)): return None
    days_diff = max(1, (pd.to_datetime(official['asof']) - pd.to_datetime(official['prev_asof'])).days)
    delta_official = official['delta_official_tonnes_vs_prev']
    prev_tonnes = official['official_tonnes'] - delta_official
    expected_fee = prev_tonnes * (GLD_ANNUAL_FEE / 365) * days_diff
    true_flow = delta_official + expected_fee
    print(f"  📊 Khoảng cách reports: {days_diff} ngày | Phí bào mòn: +{expected_fee:.3f}t")
    return {"days_between_reports": days_diff, "expected_fee_tonnes": expected_fee, "observed_change_tonnes": delta_official, "true_official_flow_tonnes": true_flow}

def calculate_fee_adjusted_flows_fixed(official, proxy):
    baseline_tonnes = official["official_tonnes"]
    proxy_tonnes = proxy["proxy_gold_tonnes"]
    official_utc = get_official_effective_utc(official["asof"])
    proxy_utc = datetime.now(timezone.utc)
    hours_elapsed = (proxy_utc - official_utc).total_seconds() / 3600
    days_elapsed = max(0, hours_elapsed / 24.0)
    expected_fee_reduction = baseline_tonnes * (GLD_ANNUAL_FEE / 365) * days_elapsed
    observed_change = proxy_tonnes - baseline_tonnes
    true_flow = observed_change + expected_fee_reduction
    print(f"  ⌛ Thời gian trôi qua: {hours_elapsed:.1f} giờ ({days_elapsed:.2f} ngày)")
    return {"observed_change_tonnes": observed_change, "fee_erosion_tonnes": expected_fee_reduction, "true_institutional_flow_tonnes": true_flow, "hours_elapsed": hours_elapsed}

# =======================
# DATA FETCHING FUNCTIONS
# =======================
def fetch_gold_spot_price():
    print("  ├─ Đang lấy giá vàng Spot (GC=F)...")
    try:
        gold_ticker = yf.Ticker("GC=F")
        hist = gold_ticker.history(period="1d")
        if not hist.empty:
            price = hist['Close'].iloc[-1]
            print(f"  ✅ Giá vàng hiện tại: ${price:,.2f}/oz")
            return float(price)
        else: raise ValueError("Yahoo Finance trả về dữ liệu rỗng")
    except Exception as e:
        print(f"  ⚠️ Lỗi lấy giá vàng: {e}. Dùng mốc dự phòng $2,680.")
        return 2680.0

def download_xlsx_bytes(url=ARCHIVE_XLSX_URL):
    print("  ├─ Đang tải XLSX archive...")
    r = requests.get(url, headers={"User-Agent": UA, "Accept": "*/*"}, timeout=90)
    content = r.content
    is_xlsx = content[:2] == b"PK"
    if (r.status_code >= 400) and (not is_xlsx): r.raise_for_status()
    if not is_xlsx: raise RuntimeError(f"Không nhận được XLSX")
    print(f"  ✅ Tải thành công ({len(content)} bytes)")
    return content, r.status_code, ""

def normalize_str(x):
    if pd.isna(x): return ""
    return re.sub(r"\s+", " ", str(x)).strip().lower()

def find_best_sheet_and_header_row(sheets: dict, scan_rows=80):
    best = (None, None, -1)
    keys = ["date", "tonne", "tonnes", "ounce", "ounces", "shares"]
    for name, df in sheets.items():
        n = min(scan_rows, len(df))
        for i in range(n):
            row = " | ".join(normalize_str(v) for v in df.iloc[i].tolist())
            score = sum(1 for k in keys if k in row)
            bonus = 0
            if "date" in row and ("tonne" in row or "ounces" in row or "ounce" in row): bonus += 2
            if "shares" in row and "outstanding" in row: bonus += 2
            score += bonus
            if score > best[2]: best = (name, i, score)
    if best[0] is None or best[2] < 2: raise RuntimeError("Không tìm được sheet/header hợp lệ")
    return best

def build_table_from_sheet(df_raw: pd.DataFrame, header_row: int):
    header = [str(x).strip() for x in df_raw.iloc[header_row].tolist()]
    df = df_raw.iloc[header_row+1:].copy()
    df.columns = header
    df = df.dropna(how="all")
    df = df.loc[:, [c for c in df.columns if normalize_str(c) not in ["", "nan", "none"]]]
    return df

def detect_col(df, patterns):
    cols = list(df.columns)
    norms = {c: normalize_str(c) for c in cols}
    for c in cols:
        s = norms[c]
        if all(p in s for p in patterns): return c
    return None

def load_official_latest():
    content, status, ctype = download_xlsx_bytes()
    print("  ├─ Đang parse XLSX...")
    sheets = pd.read_excel(io.BytesIO(content), sheet_name=None, header=None)
    sheet_name, header_row, score = find_best_sheet_and_header_row(sheets)
    print(f"  ├─ Tìm thấy sheet: '{sheet_name}' (header row {header_row})")

    df = build_table_from_sheet(sheets[sheet_name], header_row)
    date_col = detect_col(df, ["date"]) or detect_col(df, ["as", "of"]) or df.columns[0]
    tonnes_col = detect_col(df, ["tonne"])
    ounces_col = detect_col(df, ["ounce"])
    shares_col = detect_col(df, ["shares", "outstanding"])

    df2 = df.copy()
    df2[date_col] = pd.to_datetime(df2[date_col], errors="coerce")
    df2 = df2[df2[date_col].notna()].copy()
    if len(df2) == 0: raise RuntimeError("Parse date thất bại")

    def to_num(x):
        if pd.isna(x): return np.nan
        s = str(x).replace(",", "").strip()
        m = re.search(r"[-+]?\d*\.?\d+", s)
        return float(m.group(0)) if m else np.nan

    if tonnes_col: df2[tonnes_col] = df2[tonnes_col].apply(to_num)
    if ounces_col: df2[ounces_col] = df2[ounces_col].apply(to_num)
    if shares_col: df2[shares_col] = df2[shares_col].apply(to_num)
    df2 = df2.sort_values(date_col)

    last = df2.iloc[-1]
    prev = df2.iloc[-2] if len(df2) >= 2 else None
    print(f"  ✅ Lấy được dữ liệu asof: {last[date_col].date()}")

    official = {
        "asof": last[date_col].date().isoformat(),
        "official_tonnes": float(last[tonnes_col]) if tonnes_col else np.nan,
    }
    if prev is not None and tonnes_col:
        official["delta_official_tonnes_vs_prev"] = float(last[tonnes_col] - prev[tonnes_col])
        official["prev_asof"] = prev[date_col].date().isoformat()
    else:
        official["delta_official_tonnes_vs_prev"] = np.nan
        official["prev_asof"] = None
    return official

def parse_proxy_from_text(text: str):
    text = text.replace("\u00a0", " ")
    text = re.sub(r"[\s\n\r\t]+", " ", text).strip()
    m_sh = re.search(r"(?:total\s+)?shares\s+outstanding[\s:|-]*([0-9][0-9,\.]*\s*[KMB]?)", text, flags=re.I)
    m_ozps = re.search(r"ounces\s+(?:of\s+gold\s+)?per\s+share[\s:|-]*([0-9]+\.?[0-9]*)", text, flags=re.I)

    # ĐÚNG HÀM PARSE GỐC (Không dùng Lambda để tránh lỗi replace)
    def parse_human(x):
        x = x.strip()
        m = re.match(r"^([0-9][0-9,]*\.?[0-9]*)\s*([KMB])?$", x, flags=re.I)
        if not m: return np.nan
        num = float(m.group(1).replace(",", ""))
        suf = (m.group(2) or "").upper()
        mult = {"": 1.0, "K": 1e3, "M": 1e6, "B": 1e9}[suf]
        return num * mult

    shares = parse_human(m_sh.group(1)) if m_sh else np.nan
    ozps = parse_human(m_ozps.group(1)) if m_ozps else np.nan

    if not np.isfinite(shares) or not np.isfinite(ozps):
        debug_snippets = re.findall(r".{0,60}share.{0,60}", text, flags=re.I)
        for s in debug_snippets[:2]: print(f"     Debug: ...{s.strip()}...")
        if not np.isfinite(shares): raise RuntimeError("Không đọc được Total Shares Outstanding")
        if not np.isfinite(ozps): raise RuntimeError("Không đọc được Ounces of Gold Per Share")

    proxy_oz = shares * ozps
    proxy_tonnes = proxy_oz / TROY_OZ_PER_TONNE
    return float(shares), float(ozps), float(proxy_tonnes)

async def fetch_proxy_playwright():
    print("  ├─ Khởi động Chromium...")
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True, args=["--no-sandbox", "--disable-dev-shm-usage", "--disable-blink-features=AutomationControlled"])
        context = await browser.new_context(user_agent=UA, viewport={"width": 1920, "height": 1080})
        page = await context.new_page()
        try:
            await page.goto(SPDR_GLD_PAGE, wait_until="domcontentloaded", timeout=60_000)
            await page.wait_for_timeout(8000)
            try:
                await page.wait_for_function("document.body.innerText.includes('Shares Outstanding')", timeout=12000)
                print("  ✅ Tìm thấy 'Shares Outstanding' trong DOM")
            except:
                print("  ⚠️ Timeout chờ 'Shares Outstanding', nhưng vẫn tiếp tục...")
            body_text = await page.inner_text("body")
            print("  ✅ Lấy được HTML content!")
        finally:
            await browser.close()

    shares, ozps, proxy_tonnes = parse_proxy_from_text(body_text)
    return {"shares_outstanding": shares, "ounces_per_share": ozps, "proxy_gold_tonnes": proxy_tonnes}

# =======================
# MAIN EXECUTION SPDR
# =======================
print("=" * 80)
print(f"🚀 KHỞI CHẠU PHÂN TÍCH SPDR GLD (GMT+7: {datetime.now(VN_TZ).strftime('%d/%m/%Y %H:%M')})")
print("=" * 80)

print("\n[1] Lấy căn cứ giá vàng Spot...")
GOLD_PRICE = fetch_gold_spot_price()

print("\n[2] Tải dữ liệu Official GLD...")
official = load_official_latest()
thresholds, fund_value = calculate_dynamic_threshold(official['official_tonnes'], GOLD_PRICE)

print("\n[3] Scrape dữ liệu Proxy realtime...")
try:
    loop = asyncio.get_event_loop()
    proxy = loop.run_until_complete(fetch_proxy_playwright())
    proxy_success = True
except Exception as e:
    print(f"  ❌ Lỗi: {e}")
    proxy_success = False
    proxy = None

print("\n" + "=" * 80)
print("=== KẾT QUẢ PHÂN TÍCH CÓ CĂN CỨ (SPDR GLD) ===")
print("=" * 80)

report_time_vn = convert_ny_date_to_vn_str(official['asof'])

print("\n📊 OFFICIAL DATA (Historical Archive)")
print("-" * 80)
print(f"📅 Thời gian báo cáo (GMT+7): {report_time_vn}")
print(f"🏆 Official holdings: {official['official_tonnes']:,.3f} tonnes")

official_flow_analysis = calculate_official_flow_with_correct_fee(official)
if official_flow_analysis:
    flow_usd = tonnes_to_usd(official_flow_analysis['true_official_flow_tonnes'], GOLD_PRICE)
    print(f"🎯 DÒNG TIỀN OFFICIAL (Fixed): {official_flow_analysis['true_official_flow_tonnes']:+,.3f} tonnes")
    print(f"💵 Quy đổi USD: ${flow_usd:+,.0f}")
    print(f"📈 Phán đoán: {classify_flow_advanced(official_flow_analysis['true_official_flow_tonnes'], thresholds)}")

if proxy_success and proxy:
    print(f"\n📊 PROXY DATA (Realtime Scraping)")
    print("-" * 80)
    print(f"🥇 Proxy holdings: {proxy['proxy_gold_tonnes']:,.3f} tonnes")
    print(f"📊 Shares Outstanding: {proxy['shares_outstanding']:,.0f}")
    print(f"🔗 Ounces per Share: {proxy['ounces_per_share']:.6f}")

    fee_analysis = calculate_fee_adjusted_flows_fixed(official, proxy)
    flow_usd = tonnes_to_usd(fee_analysis['true_institutional_flow_tonnes'], GOLD_PRICE)

    print(f"\n🎯 DÒNG TIỀN REALTIME (Fixed): {fee_analysis['true_institutional_flow_tonnes']:+,.3f} tonnes")
    print(f"💵 Quy đổi USD: ${flow_usd:+,.0f}")
    print(f"📈 Kết luận: {classify_flow_advanced(fee_analysis['true_institutional_flow_tonnes'], thresholds)}")

    print(f"\n🔍 CHI TIẾT BREAKDOWN:")
    print(f"   ➤ Thay đổi quan sát được: {fee_analysis['observed_change_tonnes']:+,.3f} tonnes")
    print(f"   ➤ Phí bào mòn ({fee_analysis['hours_elapsed']:.1f} giờ): +{fee_analysis['fee_erosion_tonnes']:.3f} tonnes")
    print(f"   ➤ Dòng tiền institutional thực: {fee_analysis['true_institutional_flow_tonnes']:+,.3f} tonnes")
else:
    print(f"\n⚠️  PROXY DATA: Không khả dụng")

print("\n" + "=" * 80)
```

```python

import re
import os
import numpy as np
import pandas as pd
import requests
import yfinance as yf
from datetime import datetime, timezone
import pytz
import asyncio
import nest_asyncio
from playwright.async_api import async_playwright

nest_asyncio.apply()

# =======================
# CONSTANTS & SETUP
# =======================
TROY_OZ_PER_TONNE = 32150.7466
VN_TZ = pytz.timezone("Asia/Ho_Chi_Minh")
UA = "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122 Safari/537.36"
IAU_ANNUAL_FEE = 0.0025  # 0.25%

IN_COLAB = 'google.colab' in str(get_ipython())
if IN_COLAB:
    print("🔄 Đang gắn Google Drive để đồng bộ IAU...")
    from google.colab import drive
    drive.mount('/content/drive')
    DRIVE_FOLDER = "/content/drive/MyDrive/GLD_IAU_Tracker"
    os.makedirs(DRIVE_FOLDER, exist_ok=True)
    IAU_STATE_FILE = os.path.join(DRIVE_FOLDER, "iau_state.csv")
else:
    IAU_STATE_FILE = "iau_state.csv"

# =======================
# IAU CORE FUNCTIONS
# =======================
def fetch_gold_yesterday_close():
    """Lấy giá vàng chốt hôm qua (Chống nhiễu Gold Price Noise)"""
    print("  ├─ Đang lấy giá vàng chốt hôm qua...")
    try:
        hist = yf.Ticker("GC=F").history(period="2d")
        if len(hist) >= 2:
            price = float(hist['Close'].iloc[-2])
        else:
            price = float(hist['Close'].iloc[-1])
        print(f"  ✅ Giá vàng mốc chuẩn: ${price:,.2f}/oz")
        return price
    except Exception as e:
        print(f"  ⚠️ Lỗi YFinance: {e}. Dùng dự phòng $2,350.")
        return 2350.0

def fetch_gold_current_price():
    """Lấy giá vàng hiện tại (Chỉ để quy đổi ra USD cho đẹp)"""
    try:
        hist = yf.Ticker("GC=F").history(period="1d")
        if not hist.empty: return float(hist['Close'].iloc[-1])
    except: pass
    return fetch_gold_yesterday_close()

async def fetch_iau_proxy_playwright(gold_yesterday_close):
    print("  ├─ Khởi động Chromium cho IAU...")
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True, args=["--no-sandbox", "--disable-dev-shm-usage", "--disable-blink-features=AutomationControlled"])
        page = await (await browser.new_context(user_agent=UA, viewport={"width": 1920, "height": 1080})).new_page()
        try:
            await page.goto("https://www.ishares.com/us/products/239561/ishares-gold-trust", wait_until="domcontentloaded", timeout=60_000)
            await page.wait_for_timeout(10000)
            text = re.sub(r"[\s\n\r\t]+", " ", (await page.inner_text("body")).replace("\u00a0", " ")).strip()
            print("  ✅ Lấy được HTML IAU!")
        finally: await browser.close()

    m_tna = re.search(r"Net\s+Assets(?:\s+of\s+Fund)?\s+as\s+of\s+([A-Za-z]+\s+\d{1,2},\s+\d{4})[^$]*\$([0-9,]+\.?[0-9]*)", text, flags=re.I)
    if not m_tna: raise RuntimeError("Lỗi Regex IAU")

    tna_usd = float(m_tna.group(2).replace(",", ""))
    # DÙNG GIÁ HÔM QUA ĐỂ CHỐNG NHIỄU
    iau_tonnes = (tna_usd / gold_yesterday_close) / TROY_OZ_PER_TONNE
    return {"proxy_tonnes": iau_tonnes, "tna_usd": tna_usd, "report_date_raw": m_tna.group(1)}

def calculate_iau_flow_with_local_memory(current_iau_proxy):
    today_str = datetime.now(VN_TZ).strftime("%Y-%m-%d %H:%M")
    today_date_only = datetime.now(VN_TZ).strftime("%Y-%m-%d")

    # CƠ CHẾ BẢO VỆ TỰ ĐỘNG (NUKE MODE) - Chống lỗi do sửa file thủ công hoặc lỗi format
    if os.path.exists(IAU_STATE_FILE):
        try:
            df_check = pd.read_csv(IAU_STATE_FILE)
            required_cols = ['date', 'report_as_of', 'tonnes', 'tna_usd']
            if not all(col in df_check.columns for col in required_cols): raise ValueError("Thiếu cột")
            pd.to_datetime(str(df_check['date'].iloc[-1])) # Bắt lỗi thiếu năm
        except Exception as e:
            print(f"  🧹 Phát hiện file IAU bị lỗi format ({e}). Đang tự động xóa và tạo lại...")
            os.remove(IAU_STATE_FILE)

    # CHỐNG GHI TRÙNG (ANTI-SPAM) - Nếu web chưa update thì không ghi file
    if not os.path.exists(IAU_STATE_FILE):
        # ĐÃ SỬA LỖI .replace(',', '') Ở ĐÂY ĐỂ TRÁNH LỖI CSV INJECTION TỪ DẤU PHẨY TRONG NGÀY "Apr 14, 2026"
        with open(IAU_STATE_FILE, "w") as f:
            f.write("date,report_as_of,tonnes,tna_usd\n")
            f.write(f"{today_str},{current_iau_proxy['report_date_raw'].replace(',', '')},{current_iau_proxy['proxy_tonnes']},{current_iau_proxy['tna_usd']}\n")
        return None

    df_state = pd.read_csv(IAU_STATE_FILE)
    last_row = df_state.iloc[-1]

    if str(last_row['report_as_of']) == current_iau_proxy['report_date_raw'].replace(',', ''):
        print("  ⚠️ Dữ liệu web chưa cập nhật (Vẫn là báo cáo ngày cũ).")
        return None

    prev_date = str(last_row['date'])
    prev_tonnes = float(last_row['tonnes'])

    days_diff = max(1, (pd.to_datetime(today_date_only) - pd.to_datetime(prev_date)).days)
    true_flow = (current_iau_proxy['proxy_tonnes'] - prev_tonnes) + (prev_tonnes * (IAU_ANNUAL_FEE / 365) * days_diff)

    # ĐÃ SỬA LỖI .replace(',', '') Ở ĐÂY NỮA
    with open(IAU_STATE_FILE, "a") as f:
        f.write(f"{today_str},{current_iau_proxy['report_date_raw'].replace(',', '')},{current_iau_proxy['proxy_tonnes']},{current_iau_proxy['tna_usd']}\n")

    return {"true_iau_flow_tonnes": true_flow, "days_diff": days_diff, "fee_erosion": (prev_tonnes * (IAU_ANNUAL_FEE / 365) * days_diff)}

# =======================
# MAIN EXECUTION IAU
# =======================
print("=" * 80)
print(f"🚀 KHỞI CHẠU PHÂN TÍCH IAU RETAIL (GMT+7: {datetime.now(VN_TZ).strftime('%d/%m/%Y %H:%M')})")
print("=" * 80)

GOLD_YESTERDAY = fetch_gold_yesterday_close()
GOLD_CURRENT = fetch_gold_current_price()
print(f"📍 Giá vàng hiện tại: ${GOLD_CURRENT:,.2f} | Mốc tính toán: ${GOLD_YESTERDAY:,.2f}")

print("\n[1] Cào dữ liệu từ iShares...")
iau_proxy = None
try:
    loop = asyncio.get_event_loop()
    iau_proxy = loop.run_until_complete(fetch_iau_proxy_playwright(GOLD_YESTERDAY))
except Exception as e:
    print(f"  ❌ Lỗi nặng: {e}")

if iau_proxy:
    print("\n[2] Tính toán dòng tiền IAU...")
    iau_flow_data = calculate_iau_flow_with_local_memory(iau_proxy)

    print("\n" + "=" * 80)
    print("=== KẾT QUẢ PHÂN TÍCH IAU (RETAIL FLOW) ===")
    print("=" * 80)
    print(f"📦 IAU Current Tonnes: {iau_proxy['proxy_tonnes']:,.3f} tonnes")
    print(f"💵 IAU Total Net Assets: ${iau_proxy['tna_usd']/1e9:.2f} Tỷ USD")

    if iau_flow_data:
        flow_usd = iau_flow_data['true_iau_flow_tonnes'] * TROY_OZ_PER_TONNE * GOLD_CURRENT
        print("-" * 80)
        print(f"📊 Khoảng cách: {iau_flow_data['days_diff']} ngày")
        print(f"💰 Phí bào mòn (0.25%/năm): -{iau_flow_data['fee_erosion']:.3f} tonnes")
        print(f"🎯 DÒNG TIỀN THẬT IAU: {iau_flow_data['true_iau_flow_tonnes']:+.3f} tonnes (${flow_usd:+,.0f})")

        if iau_flow_data['true_iau_flow_tonnes'] > 0.1:
            print("📈 Kết luận: 🟢 RETAIL BÓP VÀO")
        elif iau_flow_data['true_iau_flow_tonnes'] < -0.1:
            print("📉 Kết luận: 🔴 RETAIL RÚT VỐN")
        else:
            print("⚪ Kết luận: BIÊN ĐỘNG NHIỄU")
    else:
        print("-" * 80)
        print("⏳ TRẠNG THÁI: Đã lưu thành công.")
        print("💡 Hãy chạy lại cell này NGÀY MAI để thấy được Dòng tiền True Flow.")
else:
    print("\n❌ Không lấy được dữ liệu IAU.")
```
